# RapidOCR

```python
import os
from docling.document_converter import DocumentConverter, PdfFormatOption
from docling.datamodel.pipeline_options import PdfPipelineOptions, RapidOcrOptions
from docling.datamodel.base_models import InputFormat


PATH = os.path.join("..","data","topik_papers","102nd-topik-test-papers","102nd-TOPIK-II-Reading-Test-Paper.pdf")

pipeline_options = PdfPipelineOptions()
pipeline_options.ocr_options = RapidOcrOptions(lang=["ko"])

converter = DocumentConverter(
    format_options={InputFormat.PDF: PdfFormatOption(pipeline_options=pipeline_options)}
)
result = converter.convert(PATH,page_range=(3,4))
document = result.document
```

```python
md = document.export_to_markdown()
```

# EasyOcr + Memory Loop

```python
from docling.datamodel.pipeline_options import PdfPipelineOptions, EasyOcrOptions
from docling.document_converter import DocumentConverter, PdfFormatOption
from docling.datamodel.base_models import InputFormat

pipeline_options = PdfPipelineOptions()
pipeline_options.ocr_options = EasyOcrOptions(lang=["ko", "en"]) # Korean + English
pipeline_options.generate_page_images = True
pipeline_options.generate_picture_images = True
pipeline_options.images_scale = 2.0 # higher resolution

results = []
chunk_size = 1
for start in range(3, 4 + 1, chunk_size):
    converter = DocumentConverter(
        format_options={InputFormat.PDF: PdfFormatOption(pipeline_options=pipeline_options)}
    )
    results.append(converter.convert(PATH,page_range=(start, start + chunk_size - 1)))
    del converter
document1 = results[0].document
document2 = results[1].document
```

# Save with images

```python
from docling_core.types.doc import ImageRefMode

# Option A: base64-embedded images (single self-contained .md file)
md = document2.export_to_markdown(image_mode=ImageRefMode.EMBEDDED)

# Option B: referenced image files (saves .png files alongside the .md)
document2.save_as_markdown("output.md", image_mode=ImageRefMode.REFERENCED)
```

```python
md = document.export_to_markdown()
```

# Layout Options

```python
from docling.datamodel.pipeline_options import PdfPipelineOptions, LayoutOptions
from docling.datamodel.layout_model_specs import DOCLING_LAYOUT_EGRET_XLARGE
from docling.document_converter import DocumentConverter, PdfFormatOption
from docling.datamodel.base_models import InputFormat

pipeline_options = PdfPipelineOptions()
pipeline_options.layout_options = LayoutOptions(model_spec=DOCLING_LAYOUT_EGRET_XLARGE)
pipeline_options.images_scale = 2.0

converter = DocumentConverter(
    format_options={InputFormat.PDF: PdfFormatOption(pipeline_options=pipeline_options)}
)
result = converter.convert(PATH,page_range=(4,4))
document = result.document
```

```python
md = document.export_to_markdown(image_mode=ImageRefMode.EMBEDDED)
```
